# Load a bundle from the public repository

This notebook discovers resources in the public GKM Starter Kit bundle repository, retrieves a bundle and its schema, and loads the bundle into GKM Python models.

## Discover available resources

```python
from ga4gh.gkm.bundles import BundleRepository

repository = BundleRepository()
repository.resource_names
```

## Retrieve a bundle and its schema

Each resource provides `bundle.json` and `bundle.schema.json`.

```python
resource_name = repository.resource_names[0]
bundle_data = repository.get_bundle(resource_name)
schema_data = repository.get_bundle_json_schema(resource_name)

resource_name, list(bundle_data), list(schema_data)[:3]
```

## Load the bundle

The repository returns parsed JSON objects. Serialize them to in-memory streams to use the standard `load_bundle` workflow.

```python
import io
import json

from ga4gh.gkm.bundles import load_bundle

bundle = load_bundle(
    io.StringIO(json.dumps(bundle_data)),
    schema=io.StringIO(json.dumps(schema_data)),
)

bundle.collection_names()
```

## Handle repository errors

An unknown resource is detected before its files are downloaded and raises `BundleRepositoryResourceNotFoundError`.

```python
from ga4gh.gkm.bundles import (
    BundleRepositoryResourceNotFoundError,
)

# 1. A resource name that is not listed in the index
try:
    repository.get_bundle("resource-that-does-not-exist")
except BundleRepositoryResourceNotFoundError as error:
    print(f"Unknown resource: {error}")
```
